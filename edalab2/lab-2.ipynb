# Lab 2: DataFrame Operations
| File | Rows | Columns | Description |
|------|------|---------|-------------|
| `bank_clients_A.csv` | 15 | 9 | Client demographics + balance |
| `bank_clients_B.csv` | 12 | 8 | Campaign contact info |

```python
import pandas as pd

df_A = pd.read_csv('bank_clients_A.csv')
df_B = pd.read_csv('bank_clients_B.csv')

print('Dataset A shape:', df_A.shape)
print('Dataset B shape:', df_B.shape)

print('\n--- Dataset A ---')
display(df_A)

print('\n--- Dataset B ---')
display(df_B)
```

## Concatenation

```python
df_concat = pd.concat([df_A, df_B], axis=0, ignore_index=True)

print('Shape after concatenation:', df_concat.shape)
print(f'  = {df_A.shape[0]} rows (A) + {df_B.shape[0]} rows (B)\n')
display(df_concat)
```

## Merging on client_id
- **Dataset A** → demographics & balance
- **Dataset B** → campaign contact info

We join on the unique key client_id to enrich each client's record.

### Inner Merge
Returns only clients whose `client_id` exists in both A and B.

```python
df_inner = pd.merge(df_A, df_B, on='client_id', how='inner')

print('Inner merge shape:', df_inner.shape)
print('client_ids kept:', df_inner['client_id'].tolist())
display(df_inner)
```

### Left Merge
Returns all rows from A; fills NaN for clients not found in B.

```python
df_left = pd.merge(df_A, df_B, on='client_id', how='left')

print('Left merge shape:', df_left.shape)
print('Clients only in A (NaN in campaign columns):', df_left[df_left['outcome'].isna()]['client_id'].tolist())
display(df_left)
```

### Outer Merge
Returns all rows from both A and B; fills NaN wherever data is missing.

```python
df_outer = pd.merge(df_A, df_B, on='client_id', how='outer')

print('Outer merge shape:', df_outer.shape)
print('client_ids only in A:', df_outer[df_outer['outcome'].isna()]['client_id'].tolist())
print('client_ids only in B:', df_outer[df_outer['balance'].isna()]['client_id'].tolist())
display(df_outer)
```
