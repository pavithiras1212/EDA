```python
import numpy as np
import pandas as pd
```

```python

height = [120, 122, 125, 127, 121, 123, 137, 131, 161, 145, 141, 132]
bins = [118, 125, 135, 160, 200]


category = pd.cut(height, bins)
print("--- Categorized Heights ---")
print(category)
print()


print("--- Bin Counts ---")
print(pd.Series(category).value_counts())
print()


bin_names = ["Short Height", "Average height", "Good Height", "Taller"]
labeled_category = pd.cut(height, bins, labels=bin_names)
print("--- Labeled Heights ---")
print(labeled_category)
print()


print("--- Equal-Length Bins (Random Data) ---")
equal_bins = pd.cut(np.random.rand(40), 5, precision=2)
print(equal_bins)
print()


randomNumbers = np.random.rand(2000)
category3 = pd.qcut(randomNumbers, 4)
print("--- Quantile Bins Counts ---")
print(pd.Series(category3).value_counts())
print("\n" + "=" * 50 + "\n")
```

```python
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/sales.csv"
df = pd.read_csv(url)


print("--- Dataset Head ---")
print(df.head(10))
print()


df["TotalPrice"] = df["UnitPrice"] * df["Quantity"]
print("--- Dataset with TotalPrice ---")
print(df.head())
print()


TotalTransaction = df["TotalPrice"]
high_value_transactions = TotalTransaction[np.abs(TotalTransaction) > 3000000]
print("--- Transactions > 3,000,000 (Count) ---")
print(f"Total matching rows: {len(high_value_transactions)}")
print(high_value_transactions.head())
print()


print("--- Outlier Rows (> 6,741,112) ---")
outliers_df = df[np.abs(TotalTransaction) > 6741112]
print(outliers_df)
```
